```python
import pandas as pd
from sklearn.metrics import precision_score, accuracy_score, recall_score, confusion_matrix, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
```

```python
iris_data = pd.read_csv("Iris.csv")
iris_data
```

```python
# convert 3 species into 0, 1, 2
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
iris_data['Species'] = le.fit_transform(iris_data['Species'])
```

```python
iris_data.sample(10)
```

```python
X = iris_data.drop(["Id","Species"], axis=1)
y = iris_data["Species"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.5, random_state=42)
```

```python
# Logistic regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
```

```python
# Naiev bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
```

```python
# KNN
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
```

```python
models = {"Logsitic Regression": lr, "Naive bayes": nb, "KNN": knn}

for name, model in models.items():
    y_pred = model.predict(X_test)
    print(f"\n{name}")
    print("Accuracy:", accuracy_score(y_test, y_pred))
    
    print("cm:", confusion_matrix(y_test, y_pred))

    print("classification report:", classification_report(y_test, y_pred))
```
